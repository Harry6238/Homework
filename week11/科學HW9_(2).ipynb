Consider the homogeneous BVP: $$u'' - 2u' + u = 0,\qquad u(0)=0,\quad u'(1)=0.$$

Find a particular solution,try constant $u_p=C,C=1$, $$u(x) = (C_1 + C_2 x)e^{x} + 1.$$
By $u(0)=0:$ $$(C_1 + 0)e^0 + 1 = 0 \implies C_1 = -1.$$
then $$u'(x) = (C_1 + C_2 + C_2 x)e^{x},$$
so for $u'(1)=1:$ $$(C_1 + 2C_2)e^{1} = 1 \quad\Rightarrow\quad C_1 + 2C_2 = e^{-1}.$$ $$2C_2 = 1 + e^{-1},\qquad C_2 = \frac{1+e^{-1}}{2}.$$
Thus the exact solution is $$\boxed{\,u(x)=\big(-1 + \tfrac{1+e^{-1}}{2}\,x\big)e^{x} + 1\,}.$$

##2-order finite-difference method




$$
u'' - 2u' + u = 1, \quad x \in [0,1]
$$
with boundary conditions:
$$
u(0) = 0, \quad u'(1) = 1$$



Let the domain be divided into $N$ equal subintervals:
$$
x_i = ih, \quad i = 0,1,\dots,N, \quad h = \frac{1}{N}
$$


Second derivative: (central difference, 2nd-order):
$$
u''(x_i) \approx \frac{u_{i-1} - 2u_i + u_{i+1}}{h^2}
$$

First derivative (central difference, 2nd-order):
$$
u'(x_i) \approx \frac{u_{i+1} - u_{i-1}}{2h}
$$


Substituting into the ODE at interior points $(i = 1, 2, \dots, N-1):$
$$
\frac{u_{i-1} - 2u_i + u_{i+1}}{h^2} - 2\left(\frac{u_{i+1} - u_{i-1}}{2h}\right) + u_i = 1
$$
Simplifying:
$$
\frac{u_{i-1} - 2u_i + u_{i+1}}{h^2} - \frac{u_{i+1} - u_{i-1}}{h} + u_i = 1
$$

Multiplying through by $h^2:$
$$
u_{i-1} - 2u_i + u_{i+1} - h(u_{i+1} - u_{i-1}) + h^2 u_i = h^2
$$

Rearranging terms:
$$
(1 + h)u_{i-1} + (-2 + h^2)u_i + (1 - h)u_{i+1} = h^2
$$

Boundary Conditions

Left boundary: $(i = 0):$

$$u_0 = 0$$

Right boundary: $(i = N):$\
 Using second-order one-sided difference:
$$
u'(1) \approx \frac{3u_N - 4u_{N-1} + u_{N-2}}{2h} = 1
$$
Multiplying through by $(2h):$
$$
3u_N - 4u_{N-1} + u_{N-2} = 2h$$

 System of Equations

For $(i = 1):$
$$
(-2 + h^2)u_1 + (1 - h)u_2 = h^2
$$

For $(i = 2, 3, \dots, N-1):$
$$
(1 + h)u_{i-1} + (-2 + h^2)u_i + (1 - h)u_{i+1} = h^2
$$

For $(i = N):$
$$u_{N-2} - 4u_{N-1} + 3u_N = 2h$$

```python
import numpy as np
import matplotlib.pyplot as plt

def exact_solution(x):
    C1 = -1.0
    C2 = 0.5 * (1.0 + np.exp(-1.0))
    return (C1 + C2 * x) * np.exp(x) + 1.0

def solve_fd(N):
    h = 1.0 / N
    M = N - 1
    a = (1.0 / h**2) + (1.0 / h)    # coeff of u_{j-1}
    b = (-2.0 / h**2) + 1.0         # coeff of u_j
    c = (1.0 / h**2) - (1.0 / h)    # coeff of u_{j+1}

    A = np.zeros((M, M))
    rhs = np.ones(M)

    for i in range(M):
        if i - 1 >= 0:
            A[i, i-1] = a
        A[i, i] = b
        if i + 1 < M:
            A[i, i+1] = c

    # Neumann at x=1: u_N = u_{N-1} + h  -> substitute into last equation
    A[M-1, M-1] += c
    rhs[M-1] -= c * h

    u_interior = np.linalg.solve(A, rhs)

    u = np.zeros(N+1)
    x = np.linspace(0.0, 1.0, N+1)
    u[0] = 0.0
    u[1:N] = u_interior
    u[N] = u[N-1] + h
    return x, u

def compute_errors(N):
    x, u_num = solve_fd(N)
    u_ex = exact_solution(x)
    err_inf = np.max(np.abs(u_num - u_ex))
    err_L2 = np.sqrt(np.sum((u_num - u_ex)**2) * (1.0/N))
    return x, u_num, u_ex, err_inf, err_L2

# run experiments
Ns = [10, 20, 40, 80, 160]
results = []
for N in Ns:
    x, u_num, u_ex, err_inf, err_L2 = compute_errors(N)
    results.append((N, 1.0/N, err_inf, err_L2))

# print table
print("  N    h         max-error       L2-approx")
for (N,h,ei,el2) in results:
    print(f"{N:4d} {h:8.5e} {ei:12.5e} {el2:12.5e}")

# compute observed rates (infinity norm)
for i in range(1, len(results)):
    Eprev = results[i-1][2]
    Ecur  = results[i][2]
    rate = np.log(Eprev/Ecur)/np.log(2)
    print(f"Rate from N={results[i-1][0]} to N={results[i][0]}: {rate:.4f}")




```
