```python
import numpy as np
import matplotlib.pyplot as plt

# Exact solutions
def exact_forward(t):
    return np.exp(-5 * t)

def exact_backward(t):
    return np.exp(5 * t)

# Forward Euler Method
def forward_euler(h):
    t = np.arange(0, 10 + h, h)
    y = np.zeros(len(t))
    y[0] = 1
    for n in range(len(t) - 1):
        y[n + 1] = y[n] * (1 - 5 * h)
    return t, y

# Backward Euler Method
def backward_euler(h):
    t = np.arange(0, 10 + h, h)
    y = np.zeros(len(t))
    y[0] = 1
    for n in range(len(t) - 1):
        y[n + 1] = y[n] / (1 - 5 * h)
    return t, y

# Step sizes
hs = [0.4, 0.41, 0.1]

# Plot Forward Euler
plt.figure(figsize=(10, 5))
t_exact = np.linspace(0, 10, 200)
plt.plot(t_exact, exact_forward(t_exact), 'k--', label='Exact Solution e^{-5t}')
for h in hs:
    t, y = forward_euler(h)
    plt.plot(t, y, 'o-', label=f'Forward Euler h={h}')
plt.title("Forward Euler: y' = -5y, y(0)=1")
plt.xlabel("t")
plt.ylabel("y(t)")
plt.legend()
plt.grid(True)
plt.show()

# Plot Backward Euler
plt.figure(figsize=(10, 5))
plt.plot(t_exact, exact_backward(t_exact), 'k--', label='Exact Solution e^{5t}')
for h in hs:
    t, y = backward_euler(h)
    plt.plot(t, y, 'o-', label=f'Backward Euler h={h}')
plt.title("Backward Euler: y' = 5y, y(0)=1")
plt.xlabel("t")
plt.ylabel("y(t)")
plt.legend()
plt.grid(True)
plt.show()
```

| Method         | Equation   | Stable when   | Behavior at h=0.4 | Behavior at h=0.41 | Behavior at h=0.1 |
| -------------- | ---------- | ------------- | ----------------- | ------------------ | ----------------- |
| Forward Euler  | ( y'=-5y ) | ( h < 0.4 )   | borderline stable | diverges           | stable & accurate |
| Backward Euler | ( y'=5y )  | always stable | overshoots growth | stable             | accurate          |
